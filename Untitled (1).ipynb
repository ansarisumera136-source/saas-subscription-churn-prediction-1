```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# Set random seed for reproducibility
np.random.seed(42)
random.seed(42)

# Generate synthetic SaaS churn data
n_customers = 5000

# Customer demographics and account info
data = {
    'customer_id': [f'CUST_{i:05d}' for i in range(1, n_customers + 1)],
    'signup_date': pd.to_datetime([
        datetime.now() - timedelta(days=random.randint(30, 1095)) 
        for _ in range(n_customers)
    ]),
    'company_size': np.random.choice(['Small (1-50)', 'Medium (51-200)', 'Large (201-1000)', 'Enterprise (1000+)'], 
                                   n_customers, p=[0.4, 0.3, 0.2, 0.1]),
    'industry': np.random.choice(['Technology', 'Healthcare', 'Finance', 'Retail', 'Manufacturing', 'Education'], 
                               n_customers, p=[0.25, 0.15, 0.15, 0.15, 0.15, 0.15]),
    'subscription_plan': np.random.choice(['Basic', 'Professional', 'Enterprise'], 
                                        n_customers, p=[0.5, 0.35, 0.15]),
    'monthly_charges': np.random.normal(150, 75),
    
    # Usage metrics
    'total_logins': np.random.poisson(25, n_customers),
    'avg_session_duration': np.random.exponential(15, n_customers),  # minutes
    'feature_usage_score': np.random.beta(2, 5, n_customers) * 100,  # 0-100 scale
    'api_calls_per_month': np.random.poisson(500, n_customers),
    'storage_usage_gb': np.random.exponential(10, n_customers),
    
    # Engagement metrics
    'support_tickets': np.random.poisson(2, n_customers),
    'days_since_last_login': np.random.exponential(5, n_customers),
    'training_sessions_attended': np.random.poisson(1, n_customers),
    'feature_adoption_rate': np.random.beta(3, 2, n_customers),
    
    # Billing and payment
    'payment_method': np.random.choice(['Credit Card', 'Bank Transfer', 'Invoice'], 
                                     n_customers, p=[0.6, 0.25, 0.15]),
    'late_payments': np.random.poisson(0.5, n_customers),
    'contract_length': np.random.choice([1, 12, 24, 36], n_customers, p=[0.3, 0.4, 0.2, 0.1]),
    'auto_renewal': np.random.choice([0, 1], n_customers, p=[0.3, 0.7]),
}

# Create DataFrame
df = pd.DataFrame(data)

# Clean and adjust data
df['monthly_charges'] = np.clip(df['monthly_charges'], 29, 999)
df['days_since_last_login'] = np.clip(df['days_since_last_login'], 0, 90)
df['avg_session_duration'] = np.clip(df['avg_session_duration'], 1, 120)
df['storage_usage_gb'] = np.clip(df['storage_usage_gb'], 0.1, 100)

# Calculate tenure in months
df['tenure_months'] = ((datetime.now() - df['signup_date']).dt.days / 30.44).round().astype(int)

# Create churn target variable with realistic logic
churn_probability = (
    0.1 +  # base probability
    (df['days_since_last_login'] > 30) * 0.3 +  # inactive users
    (df['feature_usage_score'] < 20) * 0.25 +   # low feature usage
    (df['support_tickets'] > 5) * 0.2 +         # many support issues
    (df['late_payments'] > 2) * 0.15 +          # payment issues
    (df['tenure_months'] < 3) * 0.2 +           # new customers at risk
    (df['contract_length'] == 1) * 0.1 -        # month-to-month more likely to churn
    (df['auto_renewal'] == 1) * 0.1 -           # auto-renewal reduces churn
    (df['training_sessions_attended'] > 2) * 0.15  # engaged users less likely to churn
)

# Ensure probabilities are between 0 and 1
churn_probability = np.clip(churn_probability, 0, 0.8)
df['churned'] = np.random.binomial(1, churn_probability)

# Display dataset info
print("SaaS Subscription Churn Dataset Generated")
print(f"Shape: {df.shape}")
print(f"Churn Rate: {df['churned'].mean():.2%}")
print("\nDataset Preview:")
print(df.head())
print("\nColumn Info:")
print(df.info())
print("\nChurn Distribution by Key Features:")
print(df.groupby('subscription_plan')['churned'].agg(['count', 'mean']))
```
